```python
#Conditional Statement
age=int(input("Enter Age:"))
if age>=18:
  print("Eligible for Voting")
else:
  print("Not Eligible for Voting")
```

```python
mark=int(input("Enter The Marks:"))
if mark>=90:
  print("Grade A")
elif mark>=75:
  print("Grade B")
elif mark>=60:
   print("Grade C")
else:
  print("Grade D")
```

```python
num=int(input("Enter The Number:"))
if num%2==0:
  print("Even Number")
else:
  print("Odd Number")
```

```python
year=int(input("Enter The Year:"))
if (year%4==0) and(year%400==0 or year%100!=0):
  print("Leap year")
else:
  print("Ordinary Year")
```

```python
#Nested if else:
age=int(input("Enter The Age:"))
citizen=True
if age>=18:
  if citizen:
     print("Eligible For Voting")
else:
  print("Not Eligible")
```

```python
age=int(input("Enter Thr Age:"))
if age>=18:
  if age<=60:
    print("Eligible For Driving")
  else:
    print("Not Eligible For Driving")
else:
  print("Not Eligible for diving")
```

```python
#Using Logic Operators:
#and
age=int(input("Enter The Age:"))
if age>=18 and age<=60:
  print("Can Vote")
else:
  print("Can't Vote")
```

```python
#or
age=int(input("Enter The Age:"))
if age>=18 or age>=20:
  print("Eligible for compitation")
else:
  print("Not Eligible for compitation")
```

```python
#Multiple Conditions:
salary=int(input("Enter The Salary:"))
experience=int(input("Enter The Experience:"))
if salary>50000 and experience>3:
  print("Get Promotion")
else:
  print("No Promotion")
```
